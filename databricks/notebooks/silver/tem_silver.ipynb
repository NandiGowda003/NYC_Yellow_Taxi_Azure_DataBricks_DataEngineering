```python
from pyspark.sql import functions as F

for month in [f"{m:02d}" for m in range(1, 8)]:
    path = (
        f"abfss://silver@stnyctaxide2026nandi.dfs.core.windows.net/"
        f"yellow_taxi/2026/{month}"
    )

    df = spark.read.parquet(path)

    print(f"2026-{month}: {df.count():,} records")
```

```python
year = "2026"
month = "07"
```

```python
silver_check_path = (
    f"abfss://silver@stnyctaxide2026nandi.dfs.core.windows.net/"
    f"yellow_taxi/{year}/{month}"
)

quarantine_check_path = (
    f"abfss://quarantine@stnyctaxide2026nandi.dfs.core.windows.net/"
    f"yellow_taxi/{year}/{month}"
)

df_silver_check = spark.read.parquet(silver_check_path)
df_quarantine_check = spark.read.parquet(quarantine_check_path)

print("Silver records     :", df_silver_check.count())
print("Quarantine records :", df_quarantine_check.count())
```
